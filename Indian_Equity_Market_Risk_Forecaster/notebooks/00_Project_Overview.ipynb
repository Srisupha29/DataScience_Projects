# Indian Equity Market Risk Forecaster

## Project Overview

### 1. Project Purpose

This project investigates whether machine learning and statistical
forecasting models can predict future volatility in the Indian equity
market.

The primary objective is to develop, evaluate, and validate volatility
forecasts for the NIFTY 50 market index over multiple forecasting
horizons.

Rather than predicting the exact future price of an index or individual
stock, this project focuses on forecasting market risk.

The resulting forecasts will be evaluated for their statistical
accuracy and their potential usefulness in portfolio risk management.

### 2. Research Objectives

1. Collect and validate historical Indian equity market data.
2. Construct forward-looking volatility targets.
3. Investigate market, volatility, and macroeconomic features.
4. Establish statistical forecasting benchmarks.
5. Develop and compare machine learning forecasting models.
6. Evaluate forecasting performance using time-series validation.
7. Produce probabilistic volatility forecasts.
8. Apply forecasts to portfolio risk measurement.
9. Evaluate Value at Risk (VaR) and Expected Shortfall (ES).
10. Investigate whether improved volatility forecasts provide
    practical economic value.

### 5. Forecasting Target

The primary forecasting target is forward annualized
close-to-close volatility.

Daily log returns are defined as:

r_t = ln(P_t / P_(t-1))

For a forecast horizon of h trading days:

RV(t,h) = sqrt((252/h) * sum(r_(t+j)^2))

where:

- P_t represents the adjusted closing price.
- h represents the forecast horizon.
- 252 represents the approximate number of trading days
  in a year.
- j ranges from 1 to h.

Two forecasting horizons will be investigated:

- 5 trading days: short-term volatility.
- 21 trading days: approximately one trading month.

This is a daily-return-based volatility proxy, not
intraday realized volatility.

### 6. Research Methodology

The project will follow a structured quantitative research workflow.

1. Data acquisition
   - Identify relevant market data sources.
   - Collect historical price and market indicator data.
   - Preserve raw data.

2. Data quality and exploration
   - Check missing observations and duplicate records.
   - Investigate market calendars and trading sessions.
   - Perform exploratory data analysis.

3. Target and feature engineering
   - Construct forward volatility targets.
   - Generate historical volatility and return features.
   - Incorporate relevant market indicators.

4. Statistical baselines
   - Naive forecasting.
   - Exponentially Weighted Moving Average (EWMA).
   - Heterogeneous Autoregressive (HAR) model.
   - GARCH(1,1).

5. Machine learning
   - Ridge Regression.
   - Random Forest.
   - Gradient-boosted models.

6. Forecast evaluation
   - Expanding-window walk-forward validation.
   - RMSE, MAE, and QLIKE.
   - Comparison against statistical benchmarks.

7. Probabilistic forecasting
   - Quantile forecasts.
   - Prediction intervals.
   - Forecast calibration.

8. Portfolio risk analysis
   - Portfolio volatility.
   - Value at Risk (VaR).
   - Expected Shortfall (ES).
   - Risk backtesting.

9. Economic and robustness analysis
   - Volatility-targeting strategies.
   - Transaction-cost sensitivity.
   - Stress testing.
   - Model explainability.

### 7. Research Principles

The project will follow these principles:

- No look-ahead bias.
- No random train-test splitting for time-series forecasting.
- All preprocessing and feature selection must respect
  temporal ordering.
- Hyperparameter tuning must use training data only.
- The final test period must remain untouched until
  model selection is complete.
- Forecasts must be compared against meaningful statistical
  benchmarks.
- Every experiment must be reproducible.
- Model performance must be assessed beyond a single metric.

### 8. Initial Limitations and Assumptions

- Yahoo Finance is an initial candidate data source and
  must be audited before being accepted.
- Historical NIFTY 50 constituent membership may be difficult
  to obtain consistently.
- Using only current constituents could introduce
  survivorship bias.
- Daily close-to-close volatility is a proxy for market risk
  and does not capture intraday price movements.
- Historical relationships may not remain stable across
  different market regimes.

These limitations will be documented and revisited
during the research process.

The project will be considered complete when it has:

1. A validated and reproducible market dataset.
2. Clearly defined volatility targets and features.
